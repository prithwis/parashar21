```python
client, vsID = k21.CreateVectorStore(HARIHAR, BRAHA)
print(vsID)
print(client)
```

```python
k21.TestRetrieval(client, vsID,"Venus in Virgo in the third house fine arts")
```

```python
chart = k21.read_txt(CHART)
navamsa = k21.read_txt(NAVAMSA)
dasha = k21.read_txt(DASHA)
```

```python
nativeCase = k21.buildCaseContext(chart, navamsa, dasha)
#print(nativeCase)
```

```python
question = """
Does this native show significant talent or aptitude for music,
dance, singing, drama or other fine arts?

Examine the evidence critically. Do not assume that such talent exists.
"""
```

```python
response = k21.LLM_Response(
    client,
    k21.SYSTEM_PROMPT,
    nativeCase,
    question,
    vsID
    )

print(response.output_text)
```
