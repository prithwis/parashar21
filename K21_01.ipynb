```python
!wget -O k21.py -q https://raw.githubusercontent.com/prithwis/parashar21/main/k21utils/k21.py
import k21
```

```python
# ------------------------------------------------
# required for multiple imports of the same module
# ------------------------------------------------
#import importlib
#importlib.reload(k21)
```

```python
client = k21.authenticateOpenAI()
vsID = k21.getVectorStore(client, store_name="HariharBraha")
if not(vsID) :
    print("Creating Vector Store")
    vsID = k21.createVectorStore(client,"HariharBraha",HARIHAR, BRAHA)
print(vsID)
```

```python
#k21.TestRetrieval(client, vsID,"Venus in Virgo in the third house fine arts")
```

```python
chart = k21.read_txt(CHART)
navamsa = k21.read_txt(NAVAMSA)
dasha = k21.read_txt(DASHA)
nativeData = k21.buildCaseContext(chart, navamsa, dasha)
#print(nativeCase)
```

```python
question2 = """
The horoscope is provided through three text files: chart.txt containing the birth chart, navamsa-chart.txt containing the Navamsha chart, and dasha.txt containing the Vimshottari Dasha periods.
You are also given text files containing rules and interpretations of Hindu astrology.
Use only these supplied files. Do not use any other file, URL, website, or external source of astrological information.
Treat every question as a hypothesis to be examined, not a premise to be accepted and supported.
Based on these sources, give an overall assessment...
Based on these sources, give an overall assessment of the birth chart and Navamsha chart. Identify the major strengths, weaknesses, and particularly significant features, and explain how each may manifest in the native's life.
Identify important combinations or yogas wherever supported by the sources. Where multiple sources support the same conclusion, mention this. Where the sources differ or disagree, identify the disagreement and explain how you interpret it in the context of this horoscope.
Jai Parashar.
"""
```

```python
question ="will this person be a great singer"
```

```python
model = "gpt-5.6-luna"
#model = "gpt-5.6-terra"
#model = "gpt-5.6-sol"
```

```python
response = k21.LLM_Response(
    client,
    k21.SYSTEM_PROMPT,
    nativeData,
    question,
    vsID,
    model
    )

k21.prettierPrint(response)
```
