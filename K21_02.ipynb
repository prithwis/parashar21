```python
!wget -O k21.py -q https://raw.githubusercontent.com/prithwis/parashar21/main/k21utils/k21.py
import k21
```

```python
# ------------------------------------------------
# required for multiple imports of the same module
# ------------------------------------------------
#import importlib
#importlib.reload(k21)
```

```python
client = k21.authenticateOpenAI()
vsID = k21.getVectorStore(client, store_name="HariharBraha")
if not(vsID) :
    print("Creating Vector Store")
    vsID = k21.createVectorStore(client,"HariharBraha",HARIHAR, BRAHA)
print(vsID)
```

```python
#k21.TestRetrieval(client, vsID,"Venus in Virgo in the third house fine arts")
```

```python
chart = k21.read_txt(CHART)
navamsa = k21.read_txt(NAVAMSA)
dasha = k21.read_txt(DASHA)
nativeData = k21.buildCaseContext(chart, navamsa, dasha)
#print(nativeCase)
```

```python
question = """
The horoscope is provided as computational data from Parashar21, containing the birth chart, Navamsha chart, and Vimshottari Dasha periods. You also have access, through file search, to the authorised reference corpus containing rules and interpretations of Hindu astrology.
Use only the supplied Parashar21 horoscope data and the authorised reference corpus for astrological interpretation. Do not introduce astrological rules, interpretations, yogas, significations, or other astrological knowledge from your own pretrained knowledge or from any external source.
Treat every question as a hypothesis to be examined, not a premise to be accepted and supported. Look for relevant evidence both for and against a conclusion where such evidence exists. Give greater weight to specific rules that directly match the configuration in this horoscope than to broad or general significations. Do not explain away a specific contradictory rule merely because several more general indications point in the other direction.
Based on these sources, give an overall assessment of the birth chart and Navamsha chart. Identify the major strengths, weaknesses, contradictions, and particularly significant features, and explain how each may manifest in the native's life.
Identify important combinations or yogas wherever supported by the authorised corpus. Where multiple sources support the same conclusion, mention this. Where the sources differ or disagree, identify the disagreement and explain how you interpret it in the context of this horoscope. If the available sources do not provide sufficient evidence for a conclusion, say so rather than filling the gap from general astrological knowledge.
Jai Parashar.
"""
```

```python
xquestion ="will this person be a great singer"
```

```python
model = "gpt-5.6-luna"
responseLuna = k21.LLM_Response(
    client,
    k21.SYSTEM_PROMPT,
    nativeData,
    question,
    vsID,
    model
    )

k21.prettierPrint(responseLuna)
```
