```python
client = authenticateLLM(provider=provider)
# -------------------------------------------
from google.colab import drive
drive.mount('/content/drive')
#ls /content/drive/MyDrive/Prithwis/Parashar21/K21

K21_HOME = "/content/drive/MyDrive/Prithwis/Parashar21/K21"
BRAHA   = f"{K21_HOME}/Corpus/BR_Curated_22.txt"
HARIHAR = f"{K21_HOME}/Corpus/HM_Clean_07.txt"
CHART   = f"{K21_HOME}/Natives/{id}-Chart.txt"
NAVAMSA = f"{K21_HOME}/Natives/{id}-Navamsa_Chart.txt"
DASHA   = f"{K21_HOME}/Natives/{id}-Dasha.txt"
```

```python
CONTEXT = buildContext(BRAHA, HARIHAR, CHART, NAVAMSA, DASHA)
```

```python
RoleFile = f"{K21_HOME}/Roles/khona261020.txt"
khona21 = readText(RoleFile)
```

```python
QUESTION = """
Native has written a number of fiction and non-fiction books but has not been able to find a publisher. Based on dasha antardasha dates, is there
a liklihood of public acceptance of the books and fame for the native in the next five years. First give a simple layman's answer and then back it up
with astrological logic and refernence to sources.
"""
```

```python
QUERY = buildQuery(CONTEXT, QUESTION)
```

```python
answer = callLLM(
    client=client,
    role=khona21,
    query=QUERY,
    model=model,
    provider=provider,
    effort="medium"
)
```

```python
#display(Markdown(answer))
```

```python
MD_FILE = saveAnswer(id, model, answer)
```
