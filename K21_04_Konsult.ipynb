```python
client = authenticateLLM(provider=provider)
# -------------------------------------------
from google.colab import drive
drive.mount('/content/drive')
#ls /content/drive/MyDrive/Prithwis/Parashar21/K21

K21_HOME = "/content/drive/MyDrive/Prithwis/Parashar21/K21"
BRAHA   = f"{K21_HOME}/Corpus/BR_Curated_22.txt"
HARIHAR = f"{K21_HOME}/Corpus/HM_Clean_07.txt"
CHART   = f"{K21_HOME}/Natives/{id}-Chart.txt"
NAVAMSA = f"{K21_HOME}/Natives/{id}-Navamsa_Chart.txt"
DASHA   = f"{K21_HOME}/Natives/{id}-Dasha.txt"
```

```python
CONTEXT = buildContext(BRAHA, HARIHAR, CHART, NAVAMSA, DASHA)
```

```python
RoleFile = f"{K21_HOME}/Roles/khona261020.txt"
khona21 = readText(RoleFile)
```

```python
QUESTION = """
Based on Dasha and Antardasha of the native, can you give an overview of the natives life over the next 10 years.
What would he be busy with? Will he be successful in his work? How will his health be? Will his prop3erty disputes be settled ? Will he be able to sell his ancestral property?
For each question,first give a simple laymans's answer, then justify with astrological logic
"""
```

```python
QUERY = buildQuery(CONTEXT, QUESTION)
```

```python
answer = callLLM(
    client=client,
    role=khona21,
    query=QUERY,
    model=model,
    provider=provider,
    effort="medium"
)
```
