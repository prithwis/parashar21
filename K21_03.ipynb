```python
provider = "openai"
#model = "gpt-5.6-luna"
model = "gpt-6-luna"
#model = "gpt-6-astra"

id = "QH"
```

#Access Control

```python
client = authenticateLLM(provider=provider)
# -------------------------------------------
from google.colab import drive
drive.mount('/content/drive')
#ls /content/drive/MyDrive/Prithwis/Parashar21/K21

K21_HOME = "/content/drive/MyDrive/Prithwis/Parashar21/K21"
BRAHA   = f"{K21_HOME}/Corpus/BR_Curated_22.txt"
HARIHAR = f"{K21_HOME}/Corpus/HM_Clean_07.txt"
CHART   = f"{K21_HOME}/Natives/{id}-Chart.txt"
NAVAMSA = f"{K21_HOME}/Natives/{id}-Navamsa_Chart.txt"
DASHA   = f"{K21_HOME}/Natives/{id}-Dasha.txt"
```

```python
ROLE = """
You are Khona21, a source-grounded Jyotisha interpretation system.

RULES:
1. Use ONLY the supplied Braha and Harihar reference texts.
2. Treat the supplied horoscope calculations as factual case data.
3. Do not introduce astrological rules from pretrained knowledge.
4. Identify the specific source supporting each major interpretation.
5. Examine both favourable and unfavourable indications.
6. Give priority to specific applicable rules over general significations.
7. Explicitly identify contradictions, exceptions and uncertainties.
8. Do not invent planetary positions, yogas, aspects or lordships.
9. If evidence is insufficient, say so.
"""
```

```python
QUESTION = """
What are the principal strengths and weaknesses of this horoscope?

Identify the most significant favourable and unfavourable
planetary configurations, with explanations grounded in the
supplied reference texts.

First explain your findings in simple layman's language and then justify with astrological reasoning and references to the supplied texts.

Do not make predictions about specific future events.
"""
```

```python
CONTEXT = buildContext(BRAHA, HARIHAR, CHART, NAVAMSA, DASHA)
```

```python
QUERY = buildQuery(CONTEXT, QUESTION)
```

```python
answer = callLLM(
    client,
    ROLE,
    QUERY,
    model=model,
    provider=provider,
    effort="medium"
)
```

```python
MD_FILE = saveAnswer(id, model, answer)
```
